***One Hot Encoding***

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np
import pandas as pd

data = np.array([['Red', 'Blue', 'Green', 'Red', 'Blue']])

# Convert shape from (1, 5) to (5, 1)
data = data.reshape(-1, 1)

encoder = OneHotEncoder(sparse_output=False)

encoder_data = encoder.fit_transform(data)

print(encoder_data)
```

**Bag of Words(BOW)**

```python
from sklearn.feature_extraction.text import CountVectorizer

sentences = [
    "I love AI",
    "I love Python",
    "Python is powerful"
    "I had had a cow"
]

vectorizer = CountVectorizer()

bow = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
print(bow.toarray())
```

**N-Grams**

```python
from sklearn.feature_extraction.text import CountVectorizer

sentence = ["I love machine learning"]

# Bigram
vectorizer = CountVectorizer(ngram_range=(2, 2))

ngrams = vectorizer.fit_transform(sentence)

print(vectorizer.get_feature_names_out())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

sentence = ["I love machine learning"]

vectorizer = CountVectorizer(
    ngram_range=(2, 2),
    token_pattern=r'(?u)\b\w+\b'  # Token_pattern
)

ngrams = vectorizer.fit_transform(sentence)

print(vectorizer.get_feature_names_out())
```

```python
# Trigram  → 3 words

from sklearn.feature_extraction.text import CountVectorizer

sentence = ["I love machine learning and I also Love Python Language"]

vectorizer = CountVectorizer(ngram_range=(3, 3))

ngrams = vectorizer.fit_transform(sentence)

print(vectorizer.get_feature_names_out())
```

***TF-IDF (Term Frequency – Inverse Document Frequency)***

```python
from sklearn.feature_extraction.text import TfidfVectorizer

sentences = [
    "I love Python",
    "I love machine learning",
    "Python is powerful"
]

vectorizer = TfidfVectorizer()

tfidf = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
print(tfidf.toarray())
```

```python
from gensim.models import Word2Vec

sentences = [
    ["I", "love", "machine", "learning"],
    ["I", "love", "deep", "learning"],
    ["machine", "learning", "is", "powerful"]
]

model = Word2Vec(
    sentences,
    vector_size=100,
    window=2,
    min_count=1,
    sg=1
)

# Get vector
print(model.wv["learning"])

# Find similar words
print(model.wv.most_similar("learning"))
```
