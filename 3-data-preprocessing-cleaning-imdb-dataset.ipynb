# HTML Tags Remover

```python
import re
def remove_html_tags(text):
    pattern = re.compile(r"<.*?>")
    return pattern.sub(r'',text)
```

```python
text = "<p>This is a <b>GREAT</b> product! <br> I really <i>love</i> it. <a href='https://example.com'>Click here</a></p>"
```

```python
remove_html_tags(text)
```

```python
df["review"] = df["review"].apply(remove_html_tags)
```

# Remove URL

```python
def remove_url(text):
    pattern = re.compile(r"https?://\S+|www\.\S+")
    return pattern.sub(r'',text)
```

```python
text1 = "Visit https://www.kaggle.com/code/mdnagibmahfuj/data-preprocessing-cleaning-imdb-dataset/edit."

text2 = "Check https://www.kaggle.com for datasets."

text3 = "My website is https://www.amazon.com for shopping."

text4 = "Download from https://github.com/user/project for code."
```

```python
remove_url(text1)
```

```python
remove_url(text2)
```

```python
df["review"] = df["review"].apply(remove_url)
df["review"][0]
```

# Chat_Conversion Handle

```python
chat_words = {
    "u": "you",
    "ur": "your",
    "btw": "by the way",
    "lol": "laughing out loud",
    "gr8": "great",
    "thx": "thanks",
    "omg": "oh my god"
}
```

```python
def chat_conversion(text):
    new_text = []
    for word in text.split():
        new_text.append(chat_words.get(word.lower(), word))
    return " ".join(new_text)
```

```python
text1 = "u r gr8 btw thx"

chat_conversion(text1)
```

# Stopwords

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    return " ".join([word for word in text.split() if word.lower() not in stop_words])
```

```python
df["review"] = df["review"].apply(remove_stopwords)
df["review"][0]
```

```python
df["review"] = df["review"].apply(remove_stopwords)
df["review"]
```

# Remove_emoji handle

```python
def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"
        "\U0001F300-\U0001F5FF"
        "\U0001F680-\U0001F6FF"
        "\U0001F1E0-\U0001F1FF"
        "]+", flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text)
```

```python
text1 = "This product is amazing 😍🔥 I love it ❤️"

print(remove_emoji(text1))
```

# Tokenization

```python
from nltk.tokenize import word_tokenize

def word_tokenization(text):
    return word_tokenize(text)
```

```python
from nltk.tokenize import sent_tokenize

def sentence_tokenization(text):
    return sent_tokenize(text)
```

```python
text = "This product is amazing! I love it. The quality is great."

print(word_tokenization(text))
print(sentence_tokenization(text))
```

# Stemmer

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

def stemming(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
text1 = "I am loving this product and the products are amazing"

print(stemming(text1))
```

# Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

def lemmatization(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
text1 = "The cats are playing and the dogs are running"

print(lemmatization(text1))
```
